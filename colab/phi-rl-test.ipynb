# Sample Reinforcement Learning Workflow with Phi-3.5

This code implements a basic workflow for reinforcement learning for an LLM agent running Phi-3.5.

Overall, we aim to slot this into the following workflow:

1. RAG agent (Oracle) designs a test.
2. Untrained LLM agent (Student) independently comes up with questions (or limited number)
3. Student asks the Oracle for answers to to their questions.
4. Both Oracle and Student then take the test and generates answers to the test. Oracle's answers form the Answer Key.
5. LLM agent (Grader) compares answers of the Student and the Oracle.
6. Sentiment weights are scaled from -1 to 1 corresponding to 0 correct to all correct.
7. Score on test is the reinforcement learning weight for the agent.
8. Running multiple iterations with different distributions of questions produces a heterogeneous mixture of learning weights.
9. Use those weights to do reinforcement learning.
10. Make the student do the test again, and evaluate its scores as the success metric.
